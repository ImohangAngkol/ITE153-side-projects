```python
# Challenge 5
# Guess the secret number.
import random

low = int(input("Enter the lower bound: "))
high = int(input("Enter the Higher bound: "))

def GuessNumber(low,high):
    print (f"You have 10 tries to guess the number between {low} and {high}. Start the game")

    num = random.randint(low,high)
    # total number of chances
    chances = 10
    #Guess Counter 
    gc = 0

    while gc < chances:
        gc += 1
        guess = int(input("Guess the number I'm thinking: "))
        if guess == num:
            print (f" Yes the number is {num} and guessed it in {gc} attempts!")
        elif gc >= chances and guess != num:
            print (f" Nice try but the number is {num} better luck next time!")
        elif guess > num:
            print ("Go lower")
        elif guess < num:
            print ("Go higher")
            
GuessNumber(low,high)

# https://www.geeksforgeeks.org/python/number-guessing-game-in-python/
```
